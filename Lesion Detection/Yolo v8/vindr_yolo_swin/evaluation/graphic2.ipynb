```python
import matplotlib.pyplot as plt


# ============================================================
# DATA
# ============================================================

iou = [
    0.10, 0.15, 0.20, 0.25, 0.30, 0.35,
    0.40, 0.45, 0.50, 0.55, 0.60, 0.65,
    0.70, 0.75, 0.80, 0.85, 0.90, 0.95
]

tpr = [
    0.33185840707964603,
    0.3185840707964602,
    0.3053097345132743,
    0.28982300884955753,
    0.2831858407079646,
    0.2809734513274336,
    0.2676991150442478,
    0.2588495575221239,
    0.24778761061946902,
    0.2323008849557522,
    0.21017699115044247,
    0.168141592920354,
    0.13053097345132744,
    0.09070796460176991,
    0.04424778761061947,
    0.022123893805309734,
    0.00663716814159292,
    0.0
]


# Convert to percentage
tpr_percent = [
    value * 100
    for value in tpr
]


# ============================================================
# PLOT
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    iou,
    tpr_percent,
    marker="o",
    linewidth=2,
    markersize=5,
)

plt.xlabel(
    "IoU threshold",
    fontsize=12,
)

plt.ylabel(
    "True Positive Rate (%)",
    fontsize=12,
)

plt.title(
    "TPR vs IoU threshold",
    fontsize=14,
)

plt.xticks(
    iou,
    [f"{x:.2f}" for x in iou],
    rotation=45,
)

plt.ylim(
    0,
    35,
)

plt.grid(
    True,
    alpha=0.3,
)

plt.tight_layout()

plt.savefig(
    "tpr_vs_iou.png",
    dpi=300,
    bbox_inches="tight",
)

plt.show()
```
