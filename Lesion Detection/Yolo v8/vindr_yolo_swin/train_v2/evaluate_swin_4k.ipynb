```python
import json
import matplotlib.pyplot as plt

json_path = "test_evaluation.json"

with open(json_path, "r") as f:
    data = json.load(f)

metrics = data["metrics"]

ious = [
    0.10,
    0.15,
    0.20,
    0.25,
    0.30,
    0.40,
    0.50,
]

tpr = [
    metrics[f"TPR@{iou:.2f}"]
    for iou in ious
]

plt.figure(figsize=(8, 5))

plt.plot(
    ious,
    tpr,
    marker="o",
    linewidth=2,
)

plt.xlabel("IoU threshold")
plt.ylabel("TPR")
plt.title("TPR vs IoU")

plt.xticks(ious)
plt.ylim(0, 0.5)

plt.grid(
    True,
    alpha=0.3,
)

plt.tight_layout()

plt.show()
```
